````python
# King's Cross Live Departure Board
# To pull real-time train arrivals have to switch to TFL's StopPoint API - TFL tracks every station in UK using
# a NaPTAN code - King's Cross St Pancras = 940GZZLUKSX.
import requests

# Target King's Cross using NaPTAN ID
station_id = "940GZZLUKSX"
url = f"https://api.tfl.gov.uk/StopPoint/{station_id}/Arrivals"

print("Fetching Live King's Cross Departures...\n")

response = requests.get(url)

if response.status_code == 200:
  arrivals = response.json()

  # Arrange trains by closest arrival time - TFL sends back random order based on whichever train's tracking system pings the server first - not arrival order.
  arrivals.sort(key=lambda train: train["timeToStation"])    # Instructs Python to evaluate integer number of secs left before each train reaches the platform, reorganising the dataset lowest-to-highest.

  # Print the header layout
  print(f"{'LINE' :<15}  |  {'DESTINATION' :<25}  |  {'ARRIVING IN' :<12}")    # :<15 :<25 :<12 act as tab spaces - force Python to reserve exactly that many character spaces for that piece of text - creates columns.
  print("-" * 60)

  # Loop through the top 5 closest trains
  for train in arrivals[:5]:
    line = train["lineName"]
    destination = train["destinationName"]

    # TFL gives time in seconds - convert to whole minutes
    seconds = train["timeToStation"]
    minutes = seconds // 60   # // = divide to whole number

    # Format the time display text
    if minutes == 0:
      time_text = "Due"
    else:
      time_text = f"{minutes} mins"

    # Print formatted row clearly
    print(f"{line:<15}  |  {destination:<25}  |  {time_text:<12}")

else: print("Failed to connect to the server.")


````

⬆ This code works for lines like c2c and Greater Anglia but not for the stations that are not classified as London. Therefore, I need to include the correct National Rail API.

The National Rail API is Darwin which speaks XML (Extensible Markup Language) - either need to use a community-built wrapper like OpenLDBWS or use new tool - Python library xml.etree.ElementTree or external library BeautifulSoup.

JSON = Look up data instantly using keys like train["destinationName"].

XML = Data nested in a tree of nodes (tags inside tags) - need to navigate through branches to find what you need.

The National Rail API does not use train lines, just individual stations which are kept in the database under 3-letter codes (CRS).

E.g. Benfleet = BEF

These 3-letter codes are the National Rail equivalent of a NaPTAN code.

Need to provide Python with a translator so going to use David Wheatley's open-source database of all 2,500+ UK railway stations...
